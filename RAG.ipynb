```python
model_name = "sentence-transformers/all-mpnet-base-v2"
model_kwargs = {"device": "cpu"}
encode_kwargs = {"normalize_embeddings": False}
embeddings = HuggingFaceEmbeddings(
    model_name=model_name,
    model_kwargs=model_kwargs,
    encode_kwargs=encode_kwargs,
)
```

```python
PineconeVectorStore.from_documents(texts,embeddings,index_name=os.environ["INDEX_NAME"])
```

```python
vectorstore = PineconeVectorStore(
    index_name=os.environ["INDEX_NAME"],
    embedding=embeddings,
)

retriever = vectorstore.as_retriever(search_kwargs={"k": 1})
```

```python
from langchain_openrouter import ChatOpenRouter

llm = ChatOpenRouter(model='auto')
```

```python


query = "What is a vector DB ?"


relevant_docs = retriever.invoke(query)
```

```python
print(relevant_docs)
```

```python
from langchain.tools import tool
from typing import List

@tool
def retriever_tool(query:str)->List[str]:
  """
  This tool retreives relevant docs based on query.
  Args:
    query: The query
  Returns:
    List of relevant docs
  """

  relevant_docs = retriever.invoke(query)
  return relevant_docs
```

```python
from langchain.agents import create_agent

system_prompt = "You are a RAG agent. Your job is to search from relevant docs and answer a question. You will need to use the relevant tools available to you."

tools = [retriever_tool]

```

```python
agent = create_agent(
    model=llm,
    system_prompt=system_prompt,
    tools=tools
)
```

```python
from langchain_core.messages import HumanMessage

messages = []

while True:
    user_input = input("You: ")

    if user_input.lower() in {"exit", "quit"}:
        break

    # Add the new user message
    messages.append(HumanMessage(content=user_input))

    # Invoke the agent with the full history
    result = agent.invoke({"messages": messages})

    # Update history with everything the agent produced
    # (HumanMessage, AIMessage, ToolMessage, AIMessage)
    messages = result["messages"]

    # Print only the final AI response
    print(f"AI: {messages[-1].content}")
```
